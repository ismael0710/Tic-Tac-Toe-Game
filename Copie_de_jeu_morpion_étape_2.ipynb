```python
import ipywidgets as widgets
from IPython.display import display

# VARIABLES
chiffres = [1,2,3,4,5,6,7,8,9]
mark = ''
compteur = 0

cases = ["panneau"," "," "," "," "," "," "," "," "," "]

# FONCTION VICTOIRE
def victoire():

    combinaisons = [
        (1,2,3),
        (4,5,6),
        (7,8,9),
        (1,4,7),
        (2,5,8),
        (3,6,9),
        (1,5,9),
        (3,5,7)
    ]

    for a,b,c in combinaisons:
        if cases[a] == cases[b] == cases[c] and cases[a] != " ":
            return cases[a]

    return None

# FONCTION PRINCIPALE
def verifier(chiffre):

    global compteur
    global mark

    if chiffre in chiffres:

        chiffres.remove(chiffre)

        if compteur % 2 == 0:
            mark = "X"
        else:
            mark = "O"

        cases[chiffre] = mark

        boutons[chiffre-1].description = mark

        compteur += 1

        gagnant = victoire()

        if gagnant == "X":
            resultat.value = "<h3>Joueur 1 gagne</h3>"

        elif gagnant == "O":
            resultat.value = "<h3>Joueur 2 gagne</h3>"

        elif compteur > 8:
            resultat.value = "<h3>Match Égal</h3>"

# CREATION DES BOUTONS
boutons = []

for i in range(1,10):

    bouton = widgets.Button(
        description=" ",
        layout=widgets.Layout(width='60px', height='60px')
    )

    bouton.on_click(lambda b, x=i: verifier(x))

    boutons.append(bouton)

# ORGANISATION DU PLATEAU
ligne1 = widgets.HBox([boutons[0], boutons[1], boutons[2]])
ligne2 = widgets.HBox([boutons[3], boutons[4], boutons[5]])
ligne3 = widgets.HBox([boutons[6], boutons[7], boutons[8]])

# TEXTE RESULTAT
resultat = widgets.HTML()

# AFFICHAGE
display(widgets.HTML("<h1>MORPION</h1>"))
display(ligne1, ligne2, ligne3, resultat)
```
