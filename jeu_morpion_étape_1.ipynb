```python
#  Importons les bibliothèques
import ipywidgets as widgets
from IPython.display import display
from tkinter import messagebox

#  Créons une fenêtre appelée « MORPION »
titre = widgets.HTML("<h1>MORPION</h1>")

#  Définisons une liste appelée chiffres avec les chiffres de 1 à 9
chiffres = [1, 2, 3, 4, 5, 6, 7, 8, 9]

#  Initialisons la variable mark sur une chaîne vide
mark = ''

#  Fixons la variable de comptage à 0
compteur = 0

#  Créons une liste appelée cases avec 10 éléments
cases = ["panneau", " ", " ", " ", " ", " ", " ", " ", " ", " "]

# Affichage des informations
infos = widgets.HTML( f"""
    <b>Liste chiffres :</b> {chiffres}<br>
    <b>Variable mark :</b> '{mark}'<br>
    <b>Compteur :</b> {compteur}<br>
    <b>Liste cases :</b> {cases}
    """)

# Afficher l'interface
display(titre, infos)
```
